***ITEM 1***

**1. Estudien la correlación entre el valor de la matrícula y el valor del arancel para toda la base de datos y luego separen por tipo de institución.<br>¿Los grupos se comportan parecido? ¿Vale la pena estudiar todos los grupos? Visualicen apropiadamente.**

```python
# 1. Correlación y regresión para TODA la base de datos
x_total = maule['VALOR MATRICULA (PESOS)']
y_total = maule['VALOR ARANCEL (PESOS)']

cov_total = x_total.cov(y_total)
corr_total = x_total.corr(y_total)
m_total, b_total = np.polyfit(x_total, y_total, 1)
r2_total = corr_total ** 2

print("=== TODA LA BASE DE DATOS ===")
print(f"Covarianza: {cov_total:,.1f}")
print(f"Correlación de Pearson (r): {corr_total:.2f}")
print(f"Coeficiente de determinación (R^2): {r2_total*100:.1f}%")
print(f"Modelo lineal: Arancel = {m_total:.2f} * Matricula + {b_total:.2f}\n")
```

```python
# 2. Separación por TIPO DE INSTITUCION
print("=== POR TIPO DE INSTITUCIÓN ===")
for tipo, grupo in maule.groupby('TIPO DE INSTITUCION'):
    x = grupo['VALOR MATRICULA (PESOS)']
    y = grupo['VALOR ARANCEL (PESOS)']
    r = x.corr(y)
    m, b = np.polyfit(x, y, 1)
    pct = len(grupo) / len(maule) * 100
    print(f"{tipo} (n={len(grupo)}, {pct:.1f}%): r = {r:.2f} | R^2 = {(r**2)*100:.1f}% | y = {m:.2f}x + {b:.0f}")
```

```python
# 3. Visualización apropiada: gráfico general + 4 subgráficos por institución
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
axes = axes.flatten()

for idx, (tipo, grupo) in enumerate(maule.groupby('TIPO DE INSTITUCION')):
    x = grupo['VALOR MATRICULA (PESOS)']
    y = grupo['VALOR ARANCEL (PESOS)']
    r = x.corr(y)
    m, b = np.polyfit(x, y, 1)
    
    axes[idx].scatter(x, y, alpha=0.25, s=12, color='steelblue')
    axes[idx].plot(x, m * x + b, color='red', linewidth=2, label=f'r = {r:.2f} | y = {m:.1f}x + {b:,.0f}')
    axes[idx].set_title(tipo)
    axes[idx].set_xlabel('Valor Matrícula ($)')
    axes[idx].set_ylabel('Valor Arancel ($)')
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

plt.suptitle('Relación entre Matrícula y Arancel por Tipo de Institución (Maule 2021)', fontsize=13)
plt.tight_layout()
plt.show()
```

**RESPUESTAS**

***¿Los grupos se comportan parecido?*** <br>
**No**, se observan dos patrones opuestos:<br>
- **Universidades Privadas** ($r = 0,53$; $R^2 = 28,5\%$; $m = 8,42$), <br>
- **Institutos Profesionales** ($r = 0,53$; $R^2 = 28,2\%$; $m = 2,97$)<br>
- **CFT** ($r = 0,40$; $R^2 = 16,3\%$; $m = 1,53$)<br>
Presentan una correlación positiva moderada. Destacan las Universidades Privadas, donde por cada $\$10.000$ adicionales en matrícula el arancel sube en promedio $\$84.200$.<br>
- **Universidades CRUCH** ($r = 0,13$; $R^2 = 1,6\%$; $m = 3,16$)<br>
La relación lineal es prácticamente nula, ya que cobran una matrícula casi fija por institución independientemente del costo de la carrera.

***¿Vale la pena estudiar todos los grupos?***<br>
**Sí**, por dos razones:<br>
- Ningún subsistema tiene un peso marginal (van desde $18,2\%$ en IP hasta $38,2\%$ en Universidades CRUCH). <br>
- Analizar solo el total regional distorsiona el diagnóstico al mezclar estructuras de aranceles muy distintas.

**2. ¿Cuál es la relación entre el arancel y la duración de la carrera correspondiente? ¿Es verdad que, mientras más larga la carrera, más cara? ¿Será igual en todas las áreas del conocimiento?**

```python
# 1. Relación global entre Duración Total (semestres) y Valor Arancel
var_indep = maule['DURACION TOTAL CARRERA (SEMESTRES)']
var_dep = maule['VALOR ARANCEL (PESOS)']

cov_dur = var_indep.cov(var_dep)
corr_dur = var_indep.corr(var_dep)
pendiente, intercepto = np.polyfit(var_indep, var_dep, 1)
r2_dur = corr_dur ** 2

print("=== RELACIÓN GLOBAL: DURACIÓN VS ARANCEL ===")
print(f"Covarianza: {cov_dur:,.1f}")
print(f"Correlación de Pearson (r): {corr_dur:.2f}")
print(f"Coeficiente de determinación (R^2): {r2_dur*100:.1f}%")
print(f"Modelo: Arancel = {pendiente:,.1f} * Semestres + {intercepto:,.1f}\n")

# Gráfico de dispersión global con línea de tendencia
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(var_indep, var_dep, alpha=0.2, color='green')
ax.plot(var_indep, pendiente * var_indep + intercepto, color='red',
        label=f'y = {pendiente:,.0f}x + {intercepto:,.0f} (r = {corr_dur:.2f})')
ax.set_title('Relación entre Duración Total de la Carrera y Arancel')
ax.set_xlabel('Duración Total de la Carrera (Semestres)')
ax.set_ylabel('Valor Arancel (Pesos)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

```python
# 2. Análisis separado por AREA CONOCIMIENTO
print("=== POR ÁREA DEL CONOCIMIENTO ===")
resultados_area = []
for area, grupo in maule.groupby('AREA CONOCIMIENTO'):
    x = grupo['DURACION TOTAL CARRERA (SEMESTRES)']
    y = grupo['VALOR ARANCEL (PESOS)']
    r = x.corr(y)
    m, b = np.polyfit(x, y, 1)
    resultados_area.append({
        'Área del Conocimiento': area,
        'Estudiantes (n)': len(grupo),
        'Correlación (r)': round(r, 2),
        'R^2 (%)': round((r**2)*100, 1),
        'Incremento por semestre ($)': round(m, 0)
    })

df_areas = pd.DataFrame(resultados_area).sort_values(by='Correlación (r)', ascending=False)
df_areas
```

**RESPUESTAS**

- Modelo $\text{Arancel} = 346.762 \times \text{Duración} + 57.184$ (cada semestre extra incrementa el arancel anual en $\$346.762$ en promedio)<br>
- $R^2 = 62,4\%$ (la duración explica el $62,4\%$ de la variación del arancel).

***¿Es verdad que, mientras más larga la carrera, más cara?***<br>
**Sí**, como tendencia general (en las $10$ áreas la pendiente es positiva), aunque existen excepciones puntuales en postgrados cortos de alto costo anual: por ejemplo,<br>
- La Especialización en Ortodoncia ($6$ semestres) cuesta $\$12.500.000$ anuales frente a Medicina ($14$ semestres, entre $\$6,34\text{M}$ y $\$7,45\text{M}$), y los magísteres en Administración ($3$–$4$ semestres, promedio $\$4,41\text{M}$) superan a carreras profesionales de $8$–$10$ semestres del mismo rubro ($\$2,72\text{M}$).

***¿Será igual en todas las áreas del conocimiento?***
**No**, varía tanto en intensidad ($r$) como en incremento por semestre ($m$):<br>
<br>
**Correlación muy fuerte ($r \ge 0,80$)**:<br>

3. ¿Cuál es la probabilidad de seleccionar de los estudiantes hombres de la región a uno del área de educación? ¿Cuál es la probabilidad de seleccionar de las estudiantes mujeres de la región a una del área de educación? ¿Hay una discrepancia aparente? ¿Cómo se explica?

```python
# Tabla de frecuencias absolutas por Género y Área del Conocimiento
tabla_gen_abs = pd.crosstab(maule['GENERO'], maule['AREA CONOCIMIENTO'], margins=True)
print("Frecuencias absolutas:")
print(tabla_gen_abs[['Educacion', 'All']])

# Probabilidad condicional: P(Área | Género) usando normalize='index'
tabla_gen_pct = pd.crosstab(maule['GENERO'], maule['AREA CONOCIMIENTO'], normalize='index') * 100

p_edu_hombre = tabla_gen_pct.loc['Masculino', 'Educacion']
p_edu_mujer = tabla_gen_pct.loc['Femenino', 'Educacion']

print(f"\nProbabilidad dado que es Hombre (Educación): {p_edu_hombre:.2f}%")
print(f"Probabilidad dado que es Mujer (Educación): {p_edu_mujer:.2f}%")

# Mostrar tabla completa redondeada a 1 decimal (igual que en la guía)
tabla_gen_pct.round(1)
```

**RESPUESTA**

**¿Cuál es la probabilidad de seleccionar de los estudiantes hombres de la región a uno del área de educación?**<br>
<br>
Probabilidad condicional en hombres:<br>
<br>
$P(\text{Educación} \mid \text{Masculino}) = \dfrac{1.566}{20.435} = \mathbf{7,66\%}$.

**¿Cuál es la probabilidad de seleccionar de las estudiantes mujeres de la región a una del área de educación?**<br>
<br>
Probabilidad condicional en mujeres:<br>
<br>
$P(\text{Educación} \mid \text{Femenino}) = \dfrac{5.666}{28.033} = \mathbf{20,21\%}$


**¿Hay una discrepancia aparente?¿Cómo se explica?**<br>
**Sí**, la probabilidad de que una estudiante mujer pertenezca al área de Educación es más de $2,6$ veces superior a la de un estudiante hombre ($20,21\%$ frente a $7,66\%$).<br>
<br>
Se debe a una marcada segregación vocacional por género en la elección de carreras (en Educación, el $78,35\%$ de los matriculados son mujeres y solo el $21,65\%$ son hombres), sumado a que la matrícula femenina total de la región ($28.033$) es mayor que la masculina ($20.435$). Mientras las mujeres se concentran principalmente en Salud ($33,85\%$) y Educación ($20,21\%$), los hombres se inclinan mayoritariamente hacia Tecnología ($44,12\%$).
