### 3. Con la variable Tipo de Institución, realice una tabla de frecuencias y un gráfico circular. Redacte 3 observaciones a partir de este gráfico.

```python
#Frecuencia
f_institucion = maule['TIPO DE INSTITUCION'].value_counts()

tabla_institucion = pd.DataFrame({
    'Frecuencia Absoluta(f)': f_institucion,
    'Frecuencia Relativa(h%)': f_institucion / muestra * 100
})
tabla_institucion

plt.figure(figsize=(6,6))
plt.pie(f_institucion, labels=f_institucion.index,autopct='%1.2f%%', startangle=90)
plt.title('Distribución por Tipo de institución')
plt.show()
```

**RESPUESTA**<br>
- **Observación 1 :** La principal preferencia de los estudiantes de la región del Maule son las Universidades CRUCH con 18.521 estudiantes (38.21%).<br>
<br>
- **Observación 2 :** Los Centros de Formación Técnica superan a las Universidades Privadas, demostrando un fuerte interés por la educación técnica representada con 11.957 estudiantes(24.67%).<br>
<br>
- **Observación 3:** Existe una paridad casi exacta en la cantidad de matriculados entre Institutos Profesionales y Universidades Privadas.<br>

### 4. Con la variable Valor de arancel (en pesos), realice una tabla de frecuencias y un histograma. Para la tabla y el gráfico considere 10 intervalos y redacte 3 observaciones a partir de este gráfico.

```python
#Tabla de Frecuencia 
valor_arancel = maule['VALOR ARANCEL (PESOS)']
f_arancel, intervalo_arancel = np.histogram(valor_arancel, bins=10)
tabla_arancel = pd.DataFrame({
    'Intervalo inferior': intervalo_arancel[:-1],
    'Intervalo superior': intervalo_arancel[1:],
    'Frecuencia Absoluta(f)': f_arancel,
    'Frecuencia Relativa(%)': f_arancel / muestra * 100
})
display(round(tabla_arancel, 2))

#Histograma
plt.figure(figsize=(10,6))
plt.hist(valor_arancel, bins=10, edgecolor='black')
plt.xlabel('Valor Arancel (pesos)')
plt.ylabel('Frecuencia')
plt.title('Histograma de Valor del Arancel (pesos)')
plt.show()
```

**RESPUESTA**<br>
- **Observación 1 :** La gran mayoría de los estudiantes (la moda de la distribución) se concentra en el intervalo de aranceles entre $1.250.000 y $2.500.000 con 21.738 estudiantes(44.85%).<br>
<br>
- **Observación 2 :** La distribución de los aranceles es marcadamente asimétrica hacia la derecha; es decir, los aranceles superiores a 5 millones de pesos son casos atípicos y minoritarios.<br>
<br>
- **Observación 3:** Nadie paga entre $10.000.000 y $11.250.000, pero hay un pequeño grupo (12 estudiantes) que paga los valores máximos cercanos a los 12,5 millones.<br>

### 5. Confeccione una tabla que permita responder a la pregunta ¿Hay áreas del conocimiento donde las carreras sean más caras?

```python
#Medidas de tendencia central
tabla_area = maule.groupby('AREA CONOCIMIENTO')['VALOR ARANCEL (PESOS)'].agg(
    Cantidad='count', Media='mean', Mediana='median'
).sort_values('Media', ascending=False)
display(round(tabla_area, 0))
```

**RESPUESTA**<br>
- Sí, hay diferencias en el valor de arancel según el área del conocimiento.<br>
- Salud presenta la media más alta, aproximadamente $3.610.519, seguida de Derecho con $3.504.878. Educación presenta la media más baja, aproximadamente $2.144.220.<br>
- Por lo tanto, sí existen áreas donde las carreras son más caras en promedio.

### 8. Determine los valores extremos del rango del valor de matrícula para el 35% de los estudiantes que más pagan por este concepto y redacte su respuesta.

### 10. ¿Qué influencia tiene la edad en el tipo de institución a la que ingresan los estudiantes?

```python
edad_por_institucion = maule.groupby('TIPO DE INSTITUCION')['EDAD'].agg(
    Cantidad='count', Media='mean', Mediana='median'
).sort_values('Media')
display(round(edad_por_institucion, 1))
```

**RESPUESTA**<br>
- Los promedios revelan que las Universidades (CRUCH y Privadas) atraen a un público más joven (media de 22,9 y 23,5 años respectivamente). <br>
- En cambio, los estudiantes de los CFT e Institutos Profesionales tienen una media de edad superior (25,2 y 26,3 años), lo que indica que a mayor edad, los estudiantes prefieren opciones técnicas o carreras más cortas, probablemente para compatibilizarlas con el mundo laboral.

### 11. Considerando la variable Valor de arancel (en pesos), para los estudiantes de jornada diurna y jornada vespertina, determine para cada jornada el valor de: la media, la desviación estándar y el coeficiente de variación.

```python
# Filtrar solo por jornadas Diurno y Vespertino
jornada = maule[maule['JORNADA'].isin(['Diurno', 'Vespertino'])]

# Agrupar y calcular Media y Desviación Estándar (omitiendo 'Cantidad')
tabla_jornada = jornada.groupby('JORNADA')['VALOR ARANCEL (PESOS)'].agg(
    Media='mean', 
    Desviacion_Estandar='std'
)

# Calcular el Coeficiente de Variación (CV)
tabla_jornada['Coeficiente de Variación (CV)'] = (
    tabla_jornada['Desviacion_Estandar'] / tabla_jornada['Media'] * 100
)

display(round(tabla_jornada, 2))
```

**RESPUESTA**<br>
* **Diurna**
- Media = $3.122.580 <br>
- Desviación estándar = $1.171.470<br>
- Coeficiente de variación = 37.52%.<br>
* **Vespertina**
- Media = $1.901.253<br>
- Desviación estándar = $427.836 <br>
- Coeficiente de variación = 22.50%.

### 12. En base a los cálculos obtenidos en el ítem 11, indique en cuál de ambos grupos la media del Valor de arancel es más representativa. Justifique adecuadamente su respuesta.

**RESPUESTA**<br>
- La media del valor de arancel es más representativa en la Jornada Vespertina. Esto se justifica porque su Coeficiente de Variación (22,50%) es menor que el de la Jornada Diurna (37,51%). Un menor CV indica que los valores de los aranceles están menos dispersos y más concentrados alrededor del promedio.
